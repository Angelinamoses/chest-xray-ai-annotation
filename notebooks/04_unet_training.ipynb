```python

# Cell 1: PyTorch imports and configuration

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import torch
from torch.utils.data import Dataset, DataLoader

from pathlib import Path

ROOT = Path.cwd().parent
PROCESSED_DIR = ROOT / "data" / "processed"

IMAGE_DIR = PROCESSED_DIR / "images_256"
MASK_DIR = PROCESSED_DIR / "masks_256"

TRAIN_CSV = PROCESSED_DIR / "train_manifest.csv"
VAL_CSV = PROCESSED_DIR / "validation_manifest.csv"
TEST_CSV = PROCESSED_DIR / "test_manifest.csv"

CLASS_NAMES = [
    "Aortic enlargement",
    "Cardiomegaly",
    "Pleural thickening",
    "Pulmonary fibrosis",
    "Lung Opacity"
]

IMAGE_SIZE = 256
NUM_CLASSES = len(CLASS_NAMES)

BATCH_SIZE = 8
RANDOM_STATE = 42

DEVICE = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print("PyTorch:", torch.__version__)
print("Device:", DEVICE)
print("Number of classes:", NUM_CLASSES)
print("Batch size:", BATCH_SIZE)
```

```python

# Cell 2: Load train, validation and test manifests

train_df = pd.read_csv(TRAIN_CSV)
val_df = pd.read_csv(VAL_CSV)
test_df = pd.read_csv(TEST_CSV)

print("Training:", len(train_df))
print("Validation:", len(val_df))
print("Testing:", len(test_df))

print("\nTraining manifest preview:")
display(train_df.head())
```

```python

from pathlib import Path
import zipfile

ROOT = Path(
    r"C:\Users\johnm\chest-xray-ai-annotation"
)

PROCESSED = ROOT / "data" / "processed"

FILES = [
    "images_256",
    "masks_256",
    "train_manifest.csv",
    "validation_manifest.csv",
    "test_manifest.csv"
]

ZIP_PATH = ROOT / "vindr_training_dataset.zip"

with zipfile.ZipFile(
    ZIP_PATH, "w", zipfile.ZIP_DEFLATED
) as zipf:

    for item in FILES:

        path = PROCESSED / item

        if not path.exists():
            raise FileNotFoundError(path)

        if path.is_dir():

            for file in path.glob("*.npy"):
                zipf.write(
                    file,
                    arcname=f"{item}/{file.name}"
                )

        else:
            zipf.write(path, arcname=item)

print("Dataset archive created:", ZIP_PATH)
print("Archive size:", round(ZIP_PATH.stat().st_size / (1024**2), 2), "MB")
```
